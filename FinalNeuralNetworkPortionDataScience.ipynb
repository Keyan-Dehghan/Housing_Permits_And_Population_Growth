```python
# Neural Network

import torch
import torch.nn as nn
import torch.nn.functional as F


# Create a Model Class that inherits nn.Module
class Model(nn.Module):
  # Input layer (11 features) -->
  # Hidden Layer1 -->
  # Hidden Layer2 -->
  # Hidden Layer3 -->
  # output (population growth)
    def __init__(self, in_features=15, out_features=1):
        super().__init__()  # instantiate our nn.Module
        self.out = nn.Linear(in_features, out_features)

    def forward(self, x):
        x = self.out(x)
        return x


# Pick a manual seed for randomization
torch.manual_seed(41)

# Create an instance of model
model = Model()

import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline


# Train Test Split! Set X, y

X = X_Linear
y = Housing_df['Corresponding_Population_Growth']


# Drop rows with missing values
data = pd.concat([X, y], axis=1).dropna()

X = data.drop('Corresponding_Population_Growth', axis=1)
y = data['Corresponding_Population_Growth']


# Convert these to numpy arrays
X = X.values
y = y.values
```

```python
from sklearn.model_selection import train_test_split


# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=41
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=41
)
```

```python
from sklearn.preprocessing import StandardScaler
# Standardize X features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)


# Standardize y
y_scaler = StandardScaler()

y_train = y_scaler.fit_transform(y_train.reshape(-1, 1)).flatten()
y_val = y_scaler.transform(y_val.reshape(-1, 1)).flatten()
y_test = y_scaler.transform(y_test.reshape(-1, 1)).flatten()
```

```python
# Convert X features to float tensors
X_train = torch.FloatTensor(X_train)
X_val = torch.FloatTensor(X_val)
X_test = torch.FloatTensor(X_test)


# Convert y values to float tensors
y_train = torch.FloatTensor(y_train).reshape(-1, 1)
y_val = torch.FloatTensor(y_val).reshape(-1, 1)
y_test = torch.FloatTensor(y_test).reshape(-1, 1)



# Set the criterion of model to measure the error
criterion = nn.MSELoss()

# Choose Adam Optimizer
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
# Train our model!
# Epochs? (one run thru all the training data in our network)
epochs = 1000
patience = 50

train_losses = []
val_losses = []
test_losses = []

best_val_loss = float('inf')
patience_counter = 0

for i in range(epochs):

    # Go forward and get a prediction
    y_pred = model.forward(X_train)

    # Measure the training loss/error
    loss = criterion(y_pred, y_train)

    # Keep track of training loss
    train_losses.append(loss.detach().numpy())

    # Validation loss
    with torch.no_grad():
        y_val_pred = model.forward(X_val)
        val_loss = criterion(y_val_pred, y_val)

    val_losses.append(val_loss.detach().numpy())

    # Test loss
    with torch.no_grad():
        y_test_pred = model.forward(X_test)
        test_loss = criterion(y_test_pred, y_test)

    test_losses.append(test_loss.detach().numpy())

    # Early stopping
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        patience_counter = 0
    else:
        patience_counter += 1

    if i % 5 == 0 or i == epochs - 1:
        print(f'Epoch: {i+1} | Train Loss: {loss:.4f} | Validation Loss: {val_loss:.4f} | Test Loss: {test_loss:.4f}')

    if patience_counter >= patience:
        print(f"Early stopping at epoch {i+1}")
        break

    # Back propagation
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
# Neural Network Architecture Search

import torch
import torch.nn as nn
import torch.nn.functional as F
import pandas as pd
import numpy as np
import copy

from itertools import product
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


# --------------------------------------------------
# Prepare data (same structure as your original code)
# --------------------------------------------------

X = X_Transformed
y = Housing_df['Corresponding_Population_Growth']

data = pd.concat([X, y], axis=1).dropna()

X = data.drop('Corresponding_Population_Growth', axis=1).values
y = data['Corresponding_Population_Growth'].values

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=41
)

# Train/validation split
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=41
)

# Standardize X
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

# Standardize y
y_scaler = StandardScaler()
y_train = y_scaler.fit_transform(y_train.reshape(-1, 1)).flatten()
y_val = y_scaler.transform(y_val.reshape(-1, 1)).flatten()
y_test = y_scaler.transform(y_test.reshape(-1, 1)).flatten()

# Convert to tensors
X_train = torch.FloatTensor(X_train)
X_val = torch.FloatTensor(X_val)
X_test = torch.FloatTensor(X_test)

y_train = torch.FloatTensor(y_train).reshape(-1, 1)
y_val = torch.FloatTensor(y_val).reshape(-1, 1)
y_test = torch.FloatTensor(y_test).reshape(-1, 1)


# --------------------------------------------------
# Flexible model: supports 1, 2, or 3 hidden layers
# --------------------------------------------------

class Model(nn.Module):
    def __init__(self, in_features, hidden_sizes, out_features=1):
        super().__init__()

        layers = []
        input_size = in_features

        for h in hidden_sizes:
            layers.append(nn.Linear(input_size, h))
            layers.append(nn.ReLU())
            input_size = h

        layers.append(nn.Linear(input_size, out_features))
        self.network = nn.Sequential(*layers)

    def forward(self, x):
        return self.network(x)


# --------------------------------------------------
# Train one architecture and return its best losses
# --------------------------------------------------

def train_model(hidden_sizes, epochs=1000, patience=50):
    torch.manual_seed(41)

    model = Model(
        in_features=X_train.shape[1],
        hidden_sizes=hidden_sizes
    )

    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

    best_val_loss = float('inf')
    best_state = None
    best_epoch = 0
    patience_counter = 0

    for epoch in range(epochs):

        # Training step
        model.train()
        optimizer.zero_grad()

        y_pred = model(X_train)
        loss = criterion(y_pred, y_train)

        loss.backward()
        optimizer.step()

        # Evaluate after updating weights
        model.eval()
        with torch.no_grad():
            train_loss = criterion(model(X_train), y_train).item()
            val_loss = criterion(model(X_val), y_val).item()

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())
            best_epoch = epoch + 1
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= patience:
            break

    # Restore best validation model
    model.load_state_dict(best_state)

    return model, best_val_loss, best_epoch


# --------------------------------------------------
# Search combinations of hidden-layer sizes
# --------------------------------------------------

results = []

# Set True to include all three-hidden-layer combinations.
# Three layers add 18^3 = 5,832 configurations.
test_three_layers = False

architectures = []

# One hidden layer: 18 architectures
for h1 in range(1, 19):
    architectures.append((h1,))

# Two hidden layers: 18 x 18 = 324 architectures
for h1, h2 in product(range(1, 19), repeat=2):
    architectures.append((h1, h2))

# Optional three hidden layers
if test_three_layers:
    for h1, h2, h3 in product(range(1, 19), repeat=3):
        architectures.append((h1, h2, h3))

print(f"Total architectures: {len(architectures)}")

for i, hidden_sizes in enumerate(architectures, start=1):

    model, best_val_loss, best_epoch = train_model(hidden_sizes)

    results.append({
        'hidden_sizes': hidden_sizes,
        'hidden_layers': len(hidden_sizes),
        'best_validation_loss': best_val_loss,
        'best_epoch': best_epoch
    })

    print(
        f"{i}/{len(architectures)} | "
        f"Architecture: {hidden_sizes} | "
        f"Best validation loss: {best_val_loss:.5f} | "
        f"Best epoch: {best_epoch}"
    )




# --------------------------------------------------
# Rank architectures by validation loss
# --------------------------------------------------

results_df = pd.DataFrame(results).sort_values(
    'best_validation_loss'
).reset_index(drop=True)

print("\nTop 20 architectures:")
print(results_df.head(20).to_string(index=False))

best_sizes = list(results_df.iloc[0]['hidden_sizes'])

print("\nBest architecture:", best_sizes)
print(
    "Best validation MSE:",
    round(results_df.iloc[0]['best_validation_loss'], 6)
)
print("Best epoch:", results_df.iloc[0]['best_epoch'])


# --------------------------------------------------
# Final evaluation on the test set
# --------------------------------------------------

best_model, best_val_loss, best_epoch = train_model(best_sizes)

best_model.eval()
with torch.no_grad():
    test_loss = nn.MSELoss()(best_model(X_test), y_test).item()

print("\nFinal evaluation")
print("Selected architecture:", best_sizes)
print(f"Validation MSE: {best_val_loss:.6f}")
print(f"Test MSE: {test_loss:.6f}")
```
