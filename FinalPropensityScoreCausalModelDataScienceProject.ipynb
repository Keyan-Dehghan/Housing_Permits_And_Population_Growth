```python
import numpy as np
import pandas as pd

Housing_df = pd.read_excel('/Users/keyan/Downloads/HousingPermitsandPopulationGrowth.xlsx')
pd.set_option('display.max_columns', None)
Housing_df.head()
```

```python
outside_range = (
    (Housing_df['Corresponding_Population_Growth'] < -5) |
    (Housing_df['Corresponding_Population_Growth'] > 8)
)

#This function takes a look at counties with growth levels between -5% to 8%, and the lines below say
#a. how many counties are outside the range
#b. how many counties have a population growth rate of less than -5% specifically
#c. how many counties have a population growth rate of more than 8% specifically

print("Outside range:", outside_range.sum())

print("Below -5%:", (Housing_df['Corresponding_Population_Growth'] < -5).sum())
print("Above 8%:", (Housing_df['Corresponding_Population_Growth'] > 8).sum())



outside_range_2 = (
    (Housing_df['Corresponding_Growth_In_Occupied_Housing_Units'] < -12.5) |
    (Housing_df['Corresponding_Growth_In_Occupied_Housing_Units'] > 15)
)


print("\nOutside range:", outside_range_2.sum())

print("Below -12.5%:", (Housing_df['Corresponding_Growth_In_Occupied_Housing_Units'] < -12.5).sum())
print("Above 15%:", (Housing_df['Corresponding_Growth_In_Occupied_Housing_Units'] > 15).sum())
```

```python
Housing_df = Housing_df[
    Housing_df['Corresponding_Population_Growth'].between(-5, 8)
]


Housing_df = Housing_df[
    Housing_df['Corresponding_Growth_In_Occupied_Housing_Units'].between(-12.5, 15)
]
```

```python
Housing_df['log_Total_Population'] = np.log1p(Housing_df["Total_Population"])
Housing_df['log_Median_House_Val'] = np.log1p(Housing_df["Median_House_Value"])
Housing_df['log_Foreign_Born_Pct'] = np.log1p(Housing_df["Foreign_Born_Pct"])
Housing_df['log_Population_Density'] = np.log1p(Housing_df["Population_Density"])

Housing_df["recip_Rent"] = (1 / Housing_df["Median_Rent"])
Housing_df["recip_Median_Income"] = (1 / Housing_df["Median_Household_Income"])

Housing_df["Black_Pct_squared"] = (Housing_df["Black_Pct"]**2)
Housing_df["Black_Pct_cubed"] = (Housing_df["Black_Pct"]**3)


Housing_df['White_Pct_squared'] = Housing_df['White_Pct'] ** 2
Housing_df['log_Foreign_Born_Pct'] = np.log1p(Housing_df['Foreign_Born_Pct'])
Housing_df['log_Population_Density'] = np.log1p(Housing_df['Population_Density'])

Housing_df['Permits_x_Housing_Growth'] = (
    Housing_df['Total_Permits_per_1000'] *
    Housing_df['Corresponding_Growth_In_Occupied_Housing_Units']
)

Housing_df['Permits_x_House_Value'] = (
    Housing_df['Total_Permits_per_1000'] *
    Housing_df['Median_House_Value']
)
```

```python
Housing_df = Housing_df.dropna(subset=['Corresponding_Population_Growth'])

Housing_df[Housing_df["County_State_Name"] == "Acadia Parish: Louisiana"]

#This drops any rows with an N/A (nan) value, which ends up taking a lot out
```

```python
Housing_df['Housing_Permit_Treatment'] = (Housing_df['Total_Permits_per_1000'] >= Housing_df['Total_Permits_per_1000'].median()).astype(int)
```

```python
Housing_df['Housing_Permit_Percentile'] = (Housing_df['Total_Permits_per_1000'].rank(pct=True) * 100)
```

```python
Housing_df["Housing_Permit_Decile"] = pd.qcut(
    Housing_df["Total_Permits_per_1000"],
    q=10,
    labels=False
) + 1


for i in range(1, 11):
    Housing_df[f"Permit_Decile_{i}"] = (
        Housing_df["Housing_Permit_Decile"] == i
    ).astype(int)

Housing_df.head()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors



# Treatment: at or above the 50th percentile
y = Housing_df['Housing_Permit_Treatment']

# Outcome
outcome = Housing_df['Corresponding_Population_Growth']


# Estimate Propensity Scores
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_Linear_scaled = scaler.fit_transform(X_Linear)

model_Linear = LogisticRegression(max_iter=10000)
model_Linear.fit(X_Linear_scaled, y)

Housing_df['propensity_score_Linear'] = (
    model_Linear.predict_proba(X_Linear_scaled)[:, 1]
)


# Nearest Neighbor Matching
treated = Housing_df[
    Housing_df['Housing_Permit_Treatment'] == 1
]

control = Housing_df[
    Housing_df['Housing_Permit_Treatment'] == 0
]

nn_Linear = NearestNeighbors(n_neighbors=1)

nn_Linear.fit(
    control[['propensity_score_Linear']]
)

distances, indices = nn_Linear.kneighbors(
    treated[['propensity_score_Linear']]
)

matched_control = control.iloc[indices.flatten()]

matched_data_Linear = pd.concat(
    [treated, matched_control]
)


# Estimate Treatment Effect
treated_mean = matched_data_Linear[
    matched_data_Linear['Housing_Permit_Treatment'] == 1
]['Corresponding_Population_Growth'].mean()

control_mean = matched_data_Linear[
    matched_data_Linear['Housing_Permit_Treatment'] == 0
]['Corresponding_Population_Growth'].mean()

effect_Linear = treated_mean - control_mean

print(f'Estimated Linear PSM Treatment Effect: {effect_Linear:.4f}')


# Evaluate Propensity Model
from sklearn.metrics import roc_auc_score

auc_Linear = roc_auc_score(
    y,
    model_Linear.predict_proba(X_Linear_scaled)[:, 1]
)

print(f'Linear PSM Propensity Model ROC-AUC: {auc_Linear:.4f}')
```

```python
# Standardized Mean Difference (SMD)

def calculate_smd(X, treatment):
    results = []

    for column in X.columns:
        treated = X.loc[treatment == 1, column]
        control = X.loc[treatment == 0, column]

        pooled_sd = np.sqrt(
            (treated.var() + control.var()) / 2
        )

        smd = (
            treated.mean() - control.mean()
        ) / pooled_sd

        results.append({
            'Variable': column,
            'SMD': smd,
            'Absolute_SMD': abs(smd)
        })

    return pd.DataFrame(results)


# Linear PSM: SMD before matching
smd_linear_before = calculate_smd(
    X_Linear,
    y
)

# Linear PSM: SMD after matching
smd_linear_after = calculate_smd(
    matched_data_Linear[X_Linear.columns],
    matched_data_Linear['Housing_Permit_Treatment']
)

linear_balance = smd_linear_before.merge(
    smd_linear_after,
    on='Variable',
    suffixes=('_Before', '_After')
)

pd.set_option('display.expand_frame_repr', False)
print("Linear PSM Balance")
print(linear_balance)
```

```python
import matplotlib.pyplot as plt

# Linear PSM: Covariate Balance Plot

plt.figure(figsize=(10, 6))

plt.scatter(
    linear_balance['Absolute_SMD_Before'],
    linear_balance['Variable'],
    label='Before Matching'
)

plt.scatter(
    linear_balance['Absolute_SMD_After'],
    linear_balance['Variable'],
    label='After Matching'
)

plt.axvline(
    0.10,
    linestyle='--',
    label='SMD = 0.10'
)

plt.xlabel('Absolute Standardized Mean Difference')
plt.ylabel('Covariate')
plt.title('Linear PSM Covariate Balance')
plt.legend()
plt.tight_layout()
plt.show()
```
